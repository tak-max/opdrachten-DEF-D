## Opdracht 3: Kalibratie sensor en foto

```python
# verwerk hier jullie kalibratiemetingen
data = np.zeros
data = np.array([135, 324, 490, 632, 770, 938, 1052, 1200, 1360, 1600, 1885, 2148, 2435])
afstand = np.array([5,10,15,20,25,30,35,40,45,55,65,75,85])

coefficients = np.polyfit(data, afstand, 1)

x = np.linspace(0, 3000, 1000)
y = coefficients[0] * x + coefficients[1]

plt.plot(data, afstand, 'o')
plt.plot(x, y, '-')
plt.xlabel('afstand (cm)') 
plt.ylabel('gemeten waarde (a.u.)')
plt.title('Kalibratiecurve')

print("Kalibratiecoefficients:", coefficients)
```
